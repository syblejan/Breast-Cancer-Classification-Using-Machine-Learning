```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.model_selection import GridSearchCV
from imblearn.over_sampling import SMOTE 
from scipy.stats import skew
import pickle
```

```python
%matplotlib inline
sns.set(style="whitegrid")
```

```python
# Load the data
df = pd.read_csv(r"C:\Users\user\Downloads\archive (4)\data.csv")
```

```python
# Initial inspection
print("Data Head:")
print(df.head())
print("\nData Info:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 2. Drop unnecessary columns
# 'Unnamed: 32' is full of NaN, and 'id' is irrelevant
df.drop(['Unnamed: 32', 'id'], axis=1, inplace=True, errors='ignore')
```

```python
# 3. Encode Target
# Converting M -> 1 and B -> 0
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
```

```python
# 4. Feature Selection (Using your 10 features)
top_10_features = [
    'concave points_mean', 'radius_worst', 'perimeter_worst', 'area_worst', 
    'texture_worst', 'concave points_worst', 'perimeter_mean', 'area_mean', 
    'radius_mean', 'concavity_mean'
]

X = df[top_10_features]
y = df['diagnosis']
```

```python
# 5. Scaling (Standardization)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X) # Ab saari values same range mein hain

print("Preprocessing Done! Data is now clean and scaled.")
```

```python
# Create X matrix (features) and y matrix (target)
X = df[top_10_features]  # Features matrix (X)
y = df['diagnosis']      # Target variable (y)

print(f"\n" + "="*50)
print("FEATURES AND TARGET MATRICES CREATED (10 FEATURES)")
print("="*50)
print(f"X matrix shape: {X.shape}")
print(f"y matrix shape: {y.shape}")
print(f"\nFeature names: {list(X.columns)}")
print(f"\nTarget variable 'diagnosis' distribution:")
print(y.value_counts())
print(f"\nTarget variable percentage:")
print(y.value_counts(normalize=True) * 100)
```

```python
print("\n" + "="*50)
print("CHECKING FEATURE DISTRIBUTIONS")
print("="*50)

# Feature distributions
X.hist(bins=30, figsize=(15, 12))
plt.suptitle("Feature Distributions")
plt.savefig('feature_distributions.png')

# Check skewness of each feature
print("\nSkewness of each feature (|skewness| > 1 indicates significant skew):")
skewness = X.skew()
skew_results = []
for feature, skew_val in skewness.items():
    status = '(Highly Skewed)' if abs(skew_val) > 1 else '(Acceptable)'
    print(f"{feature}: {skew_val:.4f} {status}")
    skew_results.append((feature, skew_val, status))
```

```python
# Identify skewed features
from sklearn.preprocessing import PowerTransformer
skewed_features = X.columns[abs(X.skew()) > 1]
print("Skewed features identified:", list(skewed_features))

if len(skewed_features) > 0:
    # Applying PowerTransformer
    transformer = PowerTransformer(method="yeo-johnson")
    X[skewed_features] = transformer.fit_transform(X[skewed_features])

    # Visualize transformed distributions
    X[skewed_features].hist(bins=30, figsize=(12, 10), color='skyblue', edgecolor='black')
    plt.suptitle("Transformed Feature Distributions (PowerTransformer)")
    plt.savefig('transformed_features.png')

# Check new skewness
new_skewness = X[skewed_features].skew()
print("\nNew Skewness values after transformation:")
print(new_skewness)
```

```python

# 2. Check karein ke column maujood hai
if 'diagnosis' in df.columns:
    # y matrix banayein
    y = df['diagnosis']
    
    # Class counts nikaalein
    class_counts = y.value_counts()
    print("Class counts (Raw):")
    print(class_counts)

    # 3. Bar plot banayein
    plt.figure(figsize=(6, 4))
    sns.countplot(x='diagnosis', data=df, palette=['salmon', 'skyblue'])
    plt.xlabel("Diagnosis (M = Malignant, B = Benign)")
    plt.ylabel("Number of Samples")
    plt.title("Class Distribution")
    plt.show()

    # 4. Imbalance ratio
    imbalance_ratio = class_counts.max() / class_counts.min()
    print(f"\nImbalance Ratio: {imbalance_ratio:.2f}:1")
else:
    print("Error: 'diagnosis' column nahi mila. Maujood columns ye hain:")
    print(df.columns.tolist())
```

```python
from imblearn.over_sampling import SMOTE

# 2. Check imbalance
class_counts = y.value_counts()
imbalance_ratio = class_counts.max() / class_counts.min()

print(f"Initial Imbalance ratio: {imbalance_ratio:.2f}")

# 3. Apply SMOTE only if needed
if imbalance_ratio > 1.2:
    smote = SMOTE(random_state=42)
    # Fit and resample
    X_resampled, y_resampled = smote.fit_resample(X, y)
    
    # Update X and y with balanced data
    X, y = X_resampled, y_resampled
    print("SMOTE applied successfully! Classes are now balanced.")
else:
    print("SMOTE not needed, data is already balanced.")

# 4. Visualize final class distribution
print("\nFinal Class Counts:")
print(y.value_counts())

y.value_counts().plot(kind="bar", color=['skyblue', 'salmon'])
plt.title("Class Distribution (After SMOTE Check)")
plt.xlabel("Diagnosis (0: Benign, 1: Malignant)")
plt.ylabel("Number of Samples")
plt.xticks(rotation=0)
plt.show()
```

```python
# Mapping Target
y = y.map({'M': 1, 'B': 0}) if y.dtype == 'O' else y

# train/testSplit
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# Train data par fit aur transform karein, test data ko sirf transform
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Grid Search
param_grid = {'criterion': ['gini', 'entropy' ], 'max_depth': [1, 2, 3,   ] ,'min_samples_leaf' : [2, 1]}
grid_search = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='accuracy')

# Ab ye error nahi dega
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best Score:", grid_search.best_score_)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# 1. Model ko Define karein

dt = DecisionTreeClassifier(random_state=42, max_depth=5)

# 2. Model ko Fit (Train) karein -- YEH SABSE ZAROORI HAI
dt.fit(X_train, y_train)
best_dt = grid_search.best_estimator_

# 3. Ab Predictions lein (Ab error nahi aayega)
y_train_pred = dt.predict(X_train)
y_test_pred = dt.predict(X_test)

# 4. Accuracy Check
print(f"Training Accuracy: {accuracy_score(y_train, y_train_pred)*100:.2f}%")
print(f"Test Accuracy:     {accuracy_score(y_test, y_test_pred)*100:.2f}%")

```

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
import matplotlib.pyplot as plt

# 1. Confusion Matrix Display banayein
# display_labels=['Benign', 'Malignant'] is liye taaki graph par labels clear hon
cmd = ConfusionMatrixDisplay.from_predictions(
    y_test, 
    y_test_pred, 
    display_labels=['Benign', 'Malignant'], 
    cmap='Blues',  # Blue color scheme
    xticks_rotation='horizontal'
)

# 2. Title aur labels set karein
plt.title("Confusion Matrix – Decision Tree Model")
plt.grid(False) # Grid lines hata dein taaki matrix saaf dikhe
plt.show()

# --- Bonus: Matrix ke numbers ka analysis ---
tn, fp, fn, tp = confusion_matrix(y_test, y_test_pred).ravel()
print(f"\nCorrect Predictions: {tp + tn}")
print(f"Wrong Predictions: {fp + fn}")
print(f"\nModel ne {fn} Malignant cases ko miss kiya (False Negatives).")
```
